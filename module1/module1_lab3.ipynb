The textbook lists the latest three customer ratings for GEICO, Progressive, and USAA. This notebook displays the ratings, calculates each provider's minimum, mean, and maximum, and uses those results to make a recommendation.

## 1. Enter the textbook ratings

```python
import pandas as pd

ratings = pd.DataFrame({
    "Provider": ["GEICO"] * 3 + ["Progressive"] * 3 + ["USAA"] * 3,
    "Rating": [4.7, 8.3, 9.2, 7.4, 6.7, 8.9, 3.8, 6.3, 8.1]
})
ratings
```

## 2. Summary table

```python
provider_summary = (
    ratings.groupby("Provider")["Rating"]
    .agg(Minimum="min", Mean="mean", Maximum="max")
    .round(2)
    .sort_values("Mean", ascending=False)
)
provider_summary
```

## 3. Evidence notes for your decision

- Progressive has the highest mean (7.67) and the strongest minimum (6.70), which matters if you prioritize both the typical and worst-case experience.
- GEICO has the highest maximum (9.20), but its mean is 7.40 and its minimum is 4.70.
- USAA has the lowest minimum (3.80), mean (6.07), and maximum (8.10) in this three-customer sample.

**Important:** The recommendation should acknowledge that this is a very small sample of three ratings per provider.

## 4. Recommendation

I recommend Progressive based on the three customer ratings in this sample. Progressive had the highest average rating at 7.67 and the strongest minimum rating at 6.70. The higher minimum matters because it represents the best worst-case customer experience among the three providers. GEICO reached the highest single rating at 9.20, but its average of 7.40 and minimum of 4.70 were both below Progressive's results. USAA trailed both competitors with an average of 6.07, a minimum of 3.80, and a maximum of 8.10. Although three ratings per provider form a very small sample, Progressive offers the best balance of typical performance and protection against a poor customer experience.
