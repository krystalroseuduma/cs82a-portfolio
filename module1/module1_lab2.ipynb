## 1. Load the course dataset

```python
from pathlib import Path
import pandas as pd

data_path = Path("data/chickwts.csv")
chicks = pd.read_csv(data_path)
print(f"Rows: {len(chicks)}")
print(f"Feed types: {chicks['feed'].nunique()}")
chicks.head()
```

## 2. Check the data

```python
assert len(chicks) == 71
assert set(chicks.columns) == {"weight", "feed"}
assert chicks.isna().sum().sum() == 0
print("Checks passed: 71 rows, expected columns, and no missing values.")
```

## 3. Summary table

```python
feed_summary = (
    chicks.groupby("feed")["weight"]
    .agg(Minimum="min", Mean="mean", Maximum="max")
    .round(2)
    .sort_values("Mean", ascending=False)
)
feed_summary
```
