## Step 2: Count the warm days

Count temperatures strictly above 65 degrees Fahrenheit.

```python
temps = [61, 65, 72, 58, 70, 74, 68]
warm_days = 0
for temperature in temps:
    if temperature > 65:
        warm_days = warm_days + 1
print(warm_days)
```

## Step 3: Temperature converter

```python
def to_celsius(f):
    return (f - 32) * 5 / 9

print(round(to_celsius(temps[0]), 1))
```
