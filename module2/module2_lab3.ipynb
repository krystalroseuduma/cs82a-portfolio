## Temperature cutoffs

All temperatures are in degrees Fahrenheit. My categories are: freezing at 32 or below; cold above 32 and below 50; mild from 50 up to but not including 65; warm from 65 up to but not including 80; and hot at 80 or above. These rules cover every temperature without overlaps or gaps.

## Categorize every temperature

The list includes the seven temperatures used in Lab 1, plus additional temperatures to demonstrate every category and each exact cutoff.

```python
temperatures = [61, 65, 72, 58, 70, 74, 68, 20, 32, 33, 49, 50, 64, 79, 80, 95]
print(f"{'Temperature (F)':>16}  Category")
for temperature in temperatures:
    if temperature <= 32:
        category = "freezing"
    elif temperature < 50:
        category = "cold"
    elif temperature < 65:
        category = "mild"
    elif temperature < 80:
        category = "warm"
    else:
        category = "hot"
    print(f"{temperature:>16}  {category}")
```

## Boundary check

Exactly 32 is freezing; exactly 50 is mild; exactly 65 is warm; and exactly 80 is hot. The chain checks from the lowest category upward and stops at the first true condition, so each boundary agrees with the stated rules. For example, 65 fails the mild condition (temperature < 65) and satisfies the warm condition (temperature < 80).
