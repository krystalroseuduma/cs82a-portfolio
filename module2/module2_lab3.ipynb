## Temperature cutoffs

All temperatures are in degrees Celsius. Following textbook Problem 4.4, a temperature below 15 is Cold, a temperature from 15 through 25 (inclusive) is Warm, and a temperature above 25 is Hot. These rules cover all temperatures without overlaps or gaps.

## Categorize the seven days

Use the exact weekly list provided in the textbook: 10, 16, 23, 28, 14, 22, and 30 degrees Celsius.

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]
print(f"{'Day':>3} {'Temperature (C)':>16}  Category")
for day, temperature in enumerate(temperatures, start=1):
    if temperature < 15:
        category = "Cold"
    elif temperature <= 25:
        category = "Warm"
    else:
        category = "Hot"
    print(f"{day:>3} {temperature:>16}  {category}")
```

## Boundary checks

Exactly 15 and exactly 25 are both Warm. The first condition excludes 15, while the second includes 25. Values just below 15 are Cold, and values just above 25 are Hot.

```python
boundary_cases = [(14.9, "Cold"), (15, "Warm"), (15.1, "Warm"),
                  (24.9, "Warm"), (25, "Warm"), (25.1, "Hot")]
print(f"{'Temperature (C)':>16}  {'Actual':<6} {'Expected':<8} Pass")
for temperature, expected in boundary_cases:
    if temperature < 15:
        category = "Cold"
    elif temperature <= 25:
        category = "Warm"
    else:
        category = "Hot"
    assert category == expected
    print(f"{temperature:>16}  {category:<6} {expected:<8} {category == expected}")
print("All boundary checks passed.")
```
