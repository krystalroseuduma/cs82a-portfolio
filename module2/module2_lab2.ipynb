## Multiplication by repeated addition

Both functions accept two whole numbers (nonnegative integers). Starting at zero, they add the first number once for each unit of the second number. Neither function uses the multiplication operator.

```python
def multiply_for(a, b):
    total = 0
    for count in range(b):
        total = total + a
    return total

print("For-loop result for 7 by 6:", multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    total = 0
    count = 0
    while count < b:
        total = total + a
        count = count + 1
    return total

print("While-loop result for 7 by 6:", multiply_while(7, 6))
```

## Compare both versions with the real product

The multiplication operator below is used only as a test reference, outside the two functions. Tests cover both operand orders, multiplication by zero, multiplication by one, and a repeated-addition example.

```python
test_cases = [(7, 6), (6, 7), (0, 5), (5, 0), (0, 0), (1, 9), (9, 1), (4, 3)]
print(f"{'a':>3} {'b':>3} {'for':>6} {'while':>6} {'actual':>7} {'agree':>7}")
for a, b in test_cases:
    for_result = multiply_for(a, b)
    while_result = multiply_while(a, b)
    actual = a * b
    agree = for_result == while_result == actual
    assert agree
    print(f"{a:>3} {b:>3} {for_result:>6} {while_result:>6} {actual:>7} {str(agree):>7}")
print("All test cases passed.")
```

## How the while loop stops

The counter starts at zero and increases by one after each addition. The loop runs only while the counter is less than b, so after b additions the counter equals b and the condition becomes false. When b is zero, the condition is false immediately, so no additions occur and the function correctly returns zero.
